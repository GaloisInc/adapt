```python
s = {'EVENT_MPROTECT'}

def getConcept(context,s):
    atts = frozenset(getAtts(context))
    for k in context.keys():
        if s <= context[k]:
            atts = atts & context[k]
    return atts    
concepts = {att:getConcept(context,{att}) for att in getAtts(context)}

def getExtent(context,s):
    objs = frozenset()
    for k in context.keys():
        if s <= context[k]:
            objs = objs | {k}
    return objs
{att : (concepts[att],len(getExtent(context,concepts[att]))) for att in concepts.keys()}
```

```python
def bagofwords(s,sep):
    return {w for w in s.split(sep) if w != ''}
bagofwords(" a b cde f ",' ')
```

```python
def allbagofwords(s,sep):
    bags = set()
    for x in s:
        bags = bags | bagofwords(x,sep)
    return bags

allbagofwords({'a b cde f', ' gh i jkl '}, ' ')
```

```python
context3 = {k:allbagofwords(context[k],'/') for k in context.keys()}
dictionary3 = {}
cost(context3,dictionary3)
```

```python
rePair(context3,dictionary3,thresh=5,showCost=True)
```
