groupBy turns the results of a binary map query into a grouped map.

```python
context = makeContext(getQuery(proc_eventQ),'pid','type')
context
```

getAtts gets all attributes of any object in the map.

```python
atts = getAtts(context)
atts
```


The cost function measures the cost of a (context,dictionary) pair.  The cost is naively estimated as the number of attribute occurrences times log of the total number of possible attributes, plus an additional factor for the dictionary size.  It should decrease in a RePair step, else we are increasing the description length by collapsing two attributes.


```python
cost(context,{})
```

attPairs computes all distinct pairs of attributes.

```python
pairs = attPairs(atts)
pairs
```

For a given pair of attributes, countAttPair computes the number of objects in which it appears.

```python
countAttPair(context,frozenset({'EVENT_READ','EVENT_WRITE'}))
```

countAttPairs finds the number of objects having each attribute pair

```python
attPairCounts = countAttPairs(context)
attPairCounts
```

```python
topPairs = sorted(attPairCounts, key=attPairCounts.get, reverse=True)
```

```python
topPairs[0]
```

mergePair merges a pair of attributes.  We can potentially apply this to multiple pairs in a context as long as there are no shared attributes.

```python
mergePair((context,{}),topPairs[0],'new_att1')
```

oneStepRePair finds the most common pair pattern and merges it.  This is one step of the "recursive pairing" algorithm (adjusted to sets).

```python
context = makeContext(getQuery(proc_eventQ),'pid','type')
dictionary = {}
cost(context,dictionary)
```

```python
rePair(context,dictionary)
```
